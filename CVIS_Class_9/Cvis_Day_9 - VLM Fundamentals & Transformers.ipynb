#Stopword Removal

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
stop_words = list(stopwords.words('english'))
stop_words
```

```python
def remove_stopwords(text):
    if pd.isna(text):
        return text
    words = word_tokenize(str(text))
    filtered_words = [word for word in words if word.lower() not in stop_words]
    return ' '.join(filtered_words)

df['Message'] = df['Message'].apply(remove_stopwords)
```

## word2vec

```python
from gensim.models import Word2Vec
from gensim.models.keyedvectors import KeyedVectors
```

Train Word2Vec Skip-gram model on the training data

```python
word2vec_model = Word2Vec(
    sentences=X_train,
    vector_size=100,            # Dimension of word embeddings
    window=5,                   # Context window size
    min_count=2,                # Ignore words with frequency < 2
    sg=1,                       # 1 for Skip-gram, 0 for CBOW
    workers=4,                  # Number of CPU cores
    epochs=10                   # Number of training epochs
)
```

```python
print(f"Vocabulary size: {len(word2vec_model.wv.key_to_index)}")
```

```python
# Create word to index mapping
word_index = {}
embedding_matrix = []

# Add padding token (index 0)
word_index['<PAD>'] = 0
embedding_matrix.append(np.zeros(word2vec_model.vector_size))  # Zero vector for padding

# Add unknown token (index 1)
word_index['<UNK>'] = 1
embedding_matrix.append(np.random.normal(size=word2vec_model.vector_size))  # Random vector for unknown words

# Add all words from Word2Vec vocabulary
for idx, word in enumerate(word2vec_model.wv.index_to_key, start=2):
    word_index[word] = idx
    embedding_matrix.append(word2vec_model.wv[word])

# Convert to numpy arrays
embedding_matrix = np.array(embedding_matrix)
vocab_size = len(word_index)

print(f"Final vocabulary size (with special tokens): {vocab_size}")
print(f"Embedding matrix shape: {embedding_matrix.shape}")
```

```python
print(word_index)
```

```python
def tokens_to_indices(tokens, word_index, max_len=None):
    """
    Convert list of tokens to sequence of indices
    """
    indices = []
    for token in tokens:
        if token in word_index:
            indices.append(word_index[token])
        else:
            indices.append(word_index['<UNK>'])  # Unknown word

    return indices

# Convert training tokens to sequences
X_train_indices = [tokens_to_indices(tokens, word_index) for tokens in X_train]
X_test_indices = [tokens_to_indices(tokens, word_index) for tokens in X_test]

```

```python
print(f"Sample converted sequence: {X_train[0][:10]}")  # First 10 indices
print(f"Sample converted sequence: {X_train_indices[0][:10]}")  # First 10 indices
print(f"Sample converted sequence: {X_train[1][:10]}")  # First 10 indices
print(f"Sample converted sequence: {X_train_indices[1][:10]}")  # First 10 indices
```

## Padding

Determine maximum sequence length

```python
max_len = max(len(seq) for seq in X_train_indices + X_test_indices)
max_len
# Or set a fixed maximum length
# max_len = 100  # You can choose based on your data statistics
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences

# Pad sequences
X_train_padded = pad_sequences(X_train_indices, maxlen=max_len, padding='post', truncating='post')
X_test_padded = pad_sequences(X_test_indices, maxlen=max_len, padding='post', truncating='post')
```

```python
print(f"Padded sequences shape - X_train: {X_train_padded.shape}")
print(f"Padded sequences shape - X_test: {X_test_padded.shape}")
```

## Using the Embeddings in LSTM model

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Dropout
```

```python
model = Sequential([
    #Using the word2vec embedding matrix
    Embedding(
        input_dim=vocab_size,
        output_dim=word2vec_model.vector_size,
        weights=[embedding_matrix],
        input_length=max_len,
        trainable=False  #Set to True if you want to fine-tune
    ),
    LSTM(128, dropout=0.2, recurrent_dropout=0.2),
    Dense(64, activation='relu'),
    Dropout(0.3),
    Dense(2, activation='softmax')
])

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',  # Use 'categorical_crossentropy' if y is one-hot encoded
    metrics=['accuracy']
)

print(model.summary())
```

```python
from tensorflow.keras.callbacks import EarlyStopping
```

```python
# Train the model
es = EarlyStopping(monitor="val_loss", patience=3, restore_best_weights=True)
history = model.fit(
    X_train_padded,
    y_train,
    batch_size=32,
    epochs=25,
    validation_data=(X_test_padded, y_test),
    callbacks=[es],
    verbose=1
)
```

```python
from sklearn.metrics import classification_report
import matplotlib.pyplot as plt

# Get predictions
y_pred = model.predict(X_test_padded)
y_pred_classes = np.argmax(y_pred, axis=1)

# If you used label encoding, get the original class names
class_names = label_encoder.classes_

# Generate classification report
print("Classification Report:")
print(classification_report(y_test, y_pred_classes, target_names=class_names))
```

```python
def load_glove_embeddings(glove_file_path, word2vec_output_path):
    """
    Convert GloVe format to Word2Vec format for Gensim
    """
    if not os.path.exists(word2vec_output_path):
        glove2word2vec(glove_file_path, word2vec_output_path)
        print(f"Converted GloVe to Word2Vec format: {word2vec_output_path}")
    # Load the converted model
    glove_model = KeyedVectors.load_word2vec_format(word2vec_output_path, binary=False)
    return glove_model
```

```python
def load_glove_model(glove_path):
    """
    Load GloVe embeddings manually (alternative method)
    """
    embeddings_dict = {}
    with open(glove_path, 'r', encoding='utf-8') as f:
        for line in f:
            values = line.split()
            word = values[0]
            vector = np.asarray(values[1:], dtype='float32')
            embeddings_dict[word] = vector
    return embeddings_dict
```

```python
# Load GloVe embeddings (you need to have the file downloaded)
glove_path = "/content/drive/MyDrive/glove.6B.100d.txt"  # Update with your path
glove_embeddings = load_glove_model(glove_path)
```

```python
def create_glove_embedding_matrix(word_index, glove_embeddings, embedding_dim=100):
    """
    Create embedding matrix using GloVe embeddings
    """
    vocab_size = len(word_index)
    embedding_matrix = np.zeros((vocab_size, embedding_dim))

    for word, idx in word_index.items():
        if word in glove_embeddings:
            embedding_matrix[idx] = glove_embeddings[word]
        elif word.lower() in glove_embeddings:
            embedding_matrix[idx] = glove_embeddings[word.lower()]
        else:
            # Random initialization for unknown words (excluding special tokens)
            if word not in ['<PAD>', '<UNK>']:
                embedding_matrix[idx] = np.random.normal(scale=0.6, size=(embedding_dim,))

    return embedding_matrix
```

```python
all_tokens = [token for tokens in X_train for token in tokens]
unique_words = set(all_tokens)

# Create word index
word_index = {'<PAD>': 0, '<UNK>': 1}
for idx, word in enumerate(unique_words, start=2):
    word_index[word] = idx


# Create embedding matrix
embedding_dim = 100
embedding_matrix = create_glove_embedding_matrix(word_index, glove_embeddings, embedding_dim)

print(f"Embedding matrix shape: {embedding_matrix.shape}")
```

```python
def tokens_to_indices_glove(tokens, word_index):
    """
    Convert tokens to indices for GloVe
    """
    indices = []
    for token in tokens:
        if token in word_index:
            indices.append(word_index[token])
        else:
            indices.append(word_index['<UNK>'])
    return indices

# Convert tokens to indices
X_train = [tokens_to_indices_glove(tokens, word_index) for tokens in X_train]
X_test = [tokens_to_indices_glove(tokens, word_index) for tokens in X_test]

```

```python
# Pad sequences
max_len = 100
X_train_padded = pad_sequences(X_train, maxlen=max_len, padding='post', truncating='post')
X_test_padded = pad_sequences(X_test, maxlen=max_len, padding='post', truncating='post')

print(f"X_train_padded shape: {X_train_padded.shape}")
print(f"X_test_padded shape: {X_test_padded.shape}")
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Dropout, Bidirectional

vocab_size = len(word_index)

model = Sequential([
    Embedding(
        input_dim=vocab_size,
        output_dim=embedding_dim,
        weights=[embedding_matrix],
        input_length=max_len,
        trainable=False  # Freeze GloVe embeddings
    ),
    LSTM(128, dropout=0.2, recurrent_dropout=0.2),
    Dense(64, activation='relu'),
    Dropout(0.3),
    Dense(2, activation='softmax')
])

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print(model.summary())
```
